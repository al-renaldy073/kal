# Apa Itu SVD
SVD (Singular Value Decomposition) adalah metode yang digunakan dalam aljabar linear untuk menguraikan matriks menjadi tiga matriks yang lebih sederhana, membuatnya lebih mudah untuk dianalisis dan dimanipulasi.

**Rumus Umum SVD**
Jika diberikan sebuah matriks $A$ berukuran $m \times n$, maka SVD menulisnya sebagai:
$$A=UΣV^T$$

**Dimana:**
* $A$: Matriks asli berukuran $m×n$
* $U$: Matriks ortogonal berukuran $m×m$ (vektor-vektor kolomnya disebut left **singular vectors**)
* $Σ$ Matriks diagonal berukuran $m×n$ yang berisi nilai-nilai singular (singular values)
* $V^T$: Transpos dari matriks ortogonal $V$ berukuran $n×n$ (right singular vectors)

## Kegunaan SVD
1. Reduksi Dimensi (Dimensionality Reduction)
    * SVD digunakan untuk menyederhanakan data berdimensi tinggi menjadi bentuk yang lebih ringkas tanpa kehilangan informasi penting.
    * Contoh: PCA (Principal Component Analysis) menggunakan SVD untuk mencari komponen utama data.

2. Machine Learning & Data Science
Mengurangi kompleksitas data (feature selection)
    * Digunakan dalam deteksi anomali, clustering, dan fitur ekstraksi
    * Digunakan untuk mempercepat perhitungan model pembelajaran mesin
    

3. Kompresi Gambar
    * Gambar (yang bisa dianggap sebagai matriks) bisa didekomposisi menggunakan SVD.
    * Dengan hanya menyimpan sebagian nilai singular (yang besar), gambar tetap bisa direkonstruksi dengan ukuran file yang lebih kecil, tanpa kehilangan kualitas secara signifikan.

4. Pencarian Informasi & Teks (Information Retrieval)
    * Dalam bidang NLP (Natural Language Processing), SVD digunakan dalam LSA (Latent Semantic Analysis) untuk menemukan hubungan tersembunyi antar kata dan dokumen.

5. Sistem Rekomendasi
    * SVD digunakan dalam collaborative filtering (misalnya di Netflix, YouTube, Shopee) untuk merekomendasikan item berdasarkan pola interaksi pengguna.
    * Contoh: memprediksi film yang disukai oleh pengguna berdasarkan preferensi pengguna lain.

6. Solusi Matriks dan Pseudoinvers
    * SVD membantu menyelesaikan sistem persamaan linear, terutama jika matriksnya tidak berbentuk persegi atau singular (tidak bisa dibalik).

    * Digunakan untuk menghitung pseudoinvers Moore–Penrose.

7. Pengolahan Sinyal dan Audio
    * Menghapus noise dari data sinyal atau suara

    * Digunakan dalam denoising, signal compression, dan audio separation

8. Kontrol dan Sistem Dinamik
    * Digunakan dalam kontrol sistem (control system engineering) untuk analisis kestabilan dan respons sistem



## Bagaimana formula SVD
Misalkan kita punya sebuah matriks A berukuran m × n (dengan m ≥ n atau m ≤ n), maka SVD dari A dapat dituliskan sebagai:

$$A = U \Sigma V^T$$

### Penjelasan Komponen:
1. A: Matriks asli berukuran m × n.
1. U: Matriks ortogonal berukuran m × m. Kolom-kolomnya adalah vektor eigen dari $AA^T$.
1. $Σ$ (Sigma): Matriks diagonal berukuran m × n, berisi nilai singular dari A yang diurutkan dari besar ke kecil. Nilai ini adalah akar dari eigenvalue matriks $A^TA$ atau $AA^T$.
1. V: Matriks ortogonal berukuran n × n. Kolom-kolomnya adalah vektor eigen dari $A^TA$. 
1. $V^T$: Transpos dari $V$.

### Bentuk Matriks Diagonal $Σ$:

$$\Sigma =
\begin{bmatrix}
\sigma_1 & 0 & 0 & \cdots & 0 \\
0 & \sigma_2 & 0 & \cdots & 0 \\
\vdots & \vdots & \ddots & & \vdots \\
0 & 0 & \cdots & \sigma_r & 0 \\
0 & 0 & \cdots & 0 & 0 \\
\end{bmatrix}$$

**dengan:**
* $\sigma_1 \geq \sigma_2 \geq \dots \geq \sigma_r > 0$ adalah nilai singular (positive).
* $r$ adalah rank dari matriks A.

Misal A adalah matriks 2×2:

$$A = \begin{bmatrix}
3 & 1 \\
1 & 3
\end{bmatrix}$$

Langkah:
1. Hitung eigenvalue dan eigenvector dari $ATA$ dan  $AA^T$
1. Ambil akar dari eigenvalue untuk mendapat nilai singular (isi dari Σ).
1. Bangun matriks U dan V dari eigenvector.

## Latihan
### Soal 1

1. **Contoh Sederhana:**
    Misal A adalah matriks 2×2:

    **CONTOH MATRIKS:**
    Misalkan:

    $A = \begin{bmatrix}
3 & 1 \\
1 & 3
\end{bmatrix}$

    Ukuran $A: 2×2$
    
    **LANGKAH 1: Hitung Eigenvalue dan Eigenvektor dari $A^T A$**

    Karena SVD menggunakan $A^T A$ , kita hitung dulu:

    $A^T A = \begin{bmatrix}
3 & 1 \\
1 & 3
\end{bmatrix}
\begin{bmatrix}
3 & 1 \\
1 & 3
\end{bmatrix}
= \begin{bmatrix}
10 & 6 \\
6 & 10
\end{bmatrix}$

    Lalu cari $\lambda$ dari $A^T A$:

    $det(A^T A - \lambda I) = 0$

    $\Rightarrow \left| \begin{bmatrix}
10 - \lambda & 6 \\
6 & 10 - \lambda
\end{bmatrix} \right| = 0
\Rightarrow (10 - \lambda)^2 - 36 = 0 \Rightarrow \lambda^2 - 20\lambda + 64 = 0
\Rightarrow \lambda_1 = 16, \lambda_2 = 4$

    **LANGKAH 2: Hitung Singular Value**

    $\sigma_1 = \sqrt{16} = 4, \quad \sigma_2 = \sqrt{4} = 2$

    Maka:

    $\Sigma = \begin{bmatrix}
4 & 0 \\
0 & 2
\end{bmatrix}$

    **LANGKAH 3: Hitung V dari Eigenvektor $A^T A$**

    Untuk $\lambda$ = 16:

    $(A^T A - 16 I)\mathbf{v} = 0 \Rightarrow \left[ \begin{bmatrix}
-6 & 6 \\
6 & -6
\end{bmatrix} \right] \mathbf{v} = 0 \Rightarrow v_1 = \begin{bmatrix}
1 \\
1
\end{bmatrix}$

    Untuk $\lambda$ = 4:

    $(A^T A - 4 I) \mathbf{v} = 0 \Rightarrow \left[ \begin{bmatrix}
6 & 6 \\
6 & 6
\end{bmatrix} \right] \mathbf{v} = 0 \Rightarrow v_2 = \begin{bmatrix}
-1 \\
1
\end{bmatrix}$

    Normalisasi (buat panjangnya 1):


    $\mathbf{v}_1 = \frac{1}{\sqrt{2}} \begin{bmatrix}
1 \\
1
\end{bmatrix}, \quad
\mathbf{v}_2 = \frac{1}{\sqrt{2}} \begin{bmatrix}
-1 \\
1
\end{bmatrix}$

    Jadi:

    $\mathbf{V} = \left[ \begin{bmatrix}
\frac{1}{\sqrt{2}} \\
\frac{1}{\sqrt{2}}
\end{bmatrix} \quad
\begin{bmatrix}
-\frac{1}{\sqrt{2}} \\
\frac{1}{\sqrt{2}}
\end{bmatrix} \right]$

    **LANGKAH 4: Hitung U dari rumus:**

    $u_i = \frac{1}{\sigma_i} A v_i$

    Untuk $u1$:

    $\begin{aligned}
u_1 &= \frac{1}{4} A \cdot \frac{1}{\sqrt{2}} \begin{bmatrix} 1 \\ 1 \end{bmatrix}
= \frac{1}{4 \sqrt{2}} \left[ \begin{bmatrix} 3+1 \\ 1+3 \end{bmatrix} \right]
= \frac{1}{\sqrt{2}} \begin{bmatrix} 1 \\ 1 \end{bmatrix}
\end{aligned}$

    Untuk $u2$:

    $\begin{aligned}
u_2 &= \frac{1}{2} A \cdot \frac{1}{\sqrt{2}} \begin{bmatrix} -1 \\ 1 \end{bmatrix}
= \frac{1}{2 \sqrt{2}} \left[ \begin{bmatrix} -3+1 \\ -1+3 \end{bmatrix} \right]
= \frac{1}{\sqrt{2}} \begin{bmatrix} -1 \\ 1 \end{bmatrix}
\end{aligned}$

    Jadi

    $\mathbf{U} = \left[ \begin{bmatrix}
\frac{1}{\sqrt{2}} \\
\frac{1}{\sqrt{2}}
\end{bmatrix} \quad
\begin{bmatrix}
-\frac{1}{\sqrt{2}} \\
\frac{1}{\sqrt{2}}
\end{bmatrix} \right]$

    **HASIL AKHIR: SVD dari A**

    $A = U \Sigma V^{T}$

    Dengan:
    * $\mathbf{U} = \begin{bmatrix}
\frac{1}{\sqrt{2}} & -\frac{1}{\sqrt{2}} \\
\frac{1}{\sqrt{2}} & \frac{1}{\sqrt{2}}
\end{bmatrix}$

    * $\boldsymbol{\Sigma} = \begin{bmatrix}
4 & 0 \\
0 & 2
\end{bmatrix}$

    * $\mathbf{V}^T = \begin{bmatrix}
\frac{1}{\sqrt{2}} & \frac{1}{\sqrt{2}} \\
-\frac{1}{\sqrt{2}} & \frac{1}{\sqrt{2}}
\end{bmatrix}$

### Soal 2

2. **Diberikan sebuah matriks:**

    $A = \begin{bmatrix}
3 & 7 \\
2 & 5 \\
4 & 3 \\
1 & 1
\end{bmatrix}$

    **LANGKAH 1: Hitung $A^TA$**
    
    $A^T = 
\begin{bmatrix}
3 & 2 & 4 & 1 \\
7 & 5 & 3 & 1
\end{bmatrix}$

    $A^T A = 
\begin{bmatrix}
3 & 2 & 4 & 1 \\
7 & 5 & 3 & 1
\end{bmatrix}
\begin{bmatrix}
3 & 7 \\
2 & 5 \\
4 & 3 \\
1 & 1
\end{bmatrix}$
    
    Hitung hasilnya:
    
    $A^T A = 
\begin{bmatrix}
3^2 + 2^2 + 4^2 + 1^2 & 3 \cdot 7 + 2 \cdot 5 + 4 \cdot 3 + 1 \cdot 1 \\
7 \cdot 3 + 5 \cdot 2 + 3 \cdot 4 + 1 \cdot 1 & 7^2 + 5^2 + 3^2 + 1^2
\end{bmatrix}$

    $= 
\begin{bmatrix}
9 + 4 + 16 + 1 & 21 + 10 + 12 + 1 \\
21 + 10 + 12 + 1 & 49 + 25 + 9 + 1
\end{bmatrix}
= 
\begin{bmatrix}
30 & 44 \\
44 & 84
\end{bmatrix}$

    **LANGKAH 2: Hitung eigenvalue $𝜆$ dari $A^TA$**

    $\det\left(A^T A - \lambda I\right) = 
\left|
\begin{array}{cc}
30 - \lambda & 44 \\
44 & 84 - \lambda
\end{array}
\right|
= (30 - \lambda)(84 - \lambda) - 44^2$
   
    $= \lambda^2 - 114\lambda + (30)(84) - 1936 
= \lambda^2 - 114\lambda + 2520 - 1936 
= \lambda^2 - 114\lambda + 584$
    
    Gunakan rumus kuadrat: 
    
    $\lambda = \frac{114 \pm \sqrt{114^2 - 4 \cdot 584}}{2}
= \frac{114 \pm \sqrt{12996 - 2336}}{2}
= \frac{114 \pm \sqrt{10660}}{2}$

    $\sqrt{10660} \approx 103.25 
\Rightarrow \lambda_1 \approx \frac{114 + 103.25}{2} = 108.625, 
\quad \lambda_2 \approx \frac{114 - 103.25}{2} = 5.375$

    **LANGKAH 3: Hitung singular values**
    
    $\sigma_1 = \sqrt{108.625} \approx 10.42, \quad 
\sigma_2 = \sqrt{5.375} \approx 2.32$   

    $\Rightarrow \Sigma = 
\begin{bmatrix}
10.42 & 0 \\
0 & 2.32 \\
0 & 0 \\
0 & 0
\end{bmatrix}$  

    **LANGKAH 4: Hitung eigenvektor dari $A^TA$ → matriks $V$**
    Gunakan metode substitusi atau numerik (bisa saya bantu jika ingin lanjut hitung manual), tapi secara garis besar:
    
    $V = [v_1 v_2]$, dengan $v_1 v_2$ adalah eigenvektor dari $A^TA$
    
    Setelah didapatkan $V$, bisa dihitung:
    
    **LANGKAH 5: Hitung $U$ dari rumus:**
    
    $u_i = \frac{1}{\sigma_i} A v_i \quad \text{untuk } i = 1, 2$
    
    Hasil $U$ akan berupa matriks $4 \times 2$, kemudian bisa dilengkapi ke ukuran $4 \times 2$ dengan Gram-Schmidt orthonormalization jika ingin bentuk lengkap.
    
    **LANGKAH 6: Hasil Akhir SVD**
    
    $A = U \Sigma V^T$

    * Matriks $U$ (4×4):

        $U = \begin{bmatrix}
-0.7267 & -0.3453 & -0.5739 & -0.1530 \\
-0.5124 & -0.3004 &  0.7924 &  0.1392 \\
-0.4386 &  0.8736 &  0.0817 & -0.1945 \\
-0.1306 &  0.1657 & -0.1900 &  0.9588
\end{bmatrix}$
    
    * Matriks $Σ$ (4×2):
    (Singular values diletakkan di diagonal)

         $\Sigma = \begin{bmatrix}
10.4223 & 0 \\
0 & 2.3187 \\
0 & 0 \\
0 & 0
\end{bmatrix}$
    
    * Matriks $V^T$ (2×2):

        $V^T = \begin{bmatrix}
-0.4884 & -0.8726 \\
0.8726 & -0.4884
\end{bmatrix}$

    Maka: 

    $A = U \cdot \Sigma \cdot V^T$
