# Perkalian Silang (Cross Product)
Perkalian silang (cross product) adalah operasi antara dua vektor di ruang tiga dimensi ($\mathbb{R}^{3}$) yang menghasilkan vektor baru yang:
* Tegak lurus terhadap kedua vektor asal,
* Arah ditentukan oleh aturan tangan kanan (right-hand rule),
* Panjang vektor hasilnya sebesar: $|\vec{u} \times \vec{v}| = |\vec{u}| \cdot |\vec{v}| \cdot \sin \theta$

    di mana $\theta$ adalah sudut antara $\vec{u}$ dan $\vec{v}$.

## Definisi Hasil Kali Silang
Misalkan $\vec{u} =
\begin{bmatrix}
\mathbf{u_1} \\
\mathbf{u_2} \\
\mathbf{u_3}
\end{bmatrix}
.$ dan $\vec{v} =
\begin{bmatrix}
\mathbf{v_1} \\
\mathbf{v_2} \\
\mathbf{v_3}
\end{bmatrix}$ adalah vektor-vektor di $\mathbb{R}^{3}$. 
**Hasil kali silang** dari u dan v dinotasikan sebagai $\vec{u} \times \vec{v}$, adalah vektor.

$$\vec{u} \times \vec{v} =
\begin{bmatrix}
\mathbf{u_2 v_3 - u_3v_2} \\
\mathbf{-(u_1v_3 -u_3v_1)} \\
\mathbf{u_1 v_2 - u_2v_1}
\end{bmatrix}$$

## Contoh Menghitung Hasil Kali Silang
Misalkan $\vec{u} =
\begin{bmatrix}
2 \\
-1 \\
4
\end{bmatrix}$ dan $\vec{v} =
\begin{bmatrix}
3 \\
2 \\
5
\end{bmatrix}$ temukan $\vec{u} \times \vec{v}$, dan verifikasi bahwa hasil kali silang ini tegak lurus terhadap baik $\vec{u}$ maupun $\vec{v}$.

Menggunakan **Definisi**

$\vec{u} \times \vec{v} =
\begin{bmatrix}
\mathbf{u_2 v_3 - u_3v_2} \\
\mathbf{-(u_1v_3-u_3v_1)} \\
\mathbf{u_1 v_2 - u_2v_1}
\end{bmatrix} =
\begin{bmatrix}
(-1)(5) - (4)(2) \\
-((2)(5) - (4)(3))\\
(2)(2) - (-1)(3)
\end{bmatrix} =
\begin{bmatrix}
-5 -8 \\
-(10-12)\\
4+3
\end{bmatrix}=
\begin{bmatrix}
-13 \\
2\\
7
\end{bmatrix}$

**Jadi:**

$$\vec{u} \times \vec{v} = \begin{bmatrix}
-13 \\
2\\
7
\end{bmatrix}$$

**Verifikasi: Tegak Lurus**
Dua vektor tegak lurus jika hasil dot product-nya nol:
1. Cek apakah $\vec{u} \cdot (\vec{u} \times \vec{v}) = 0$:

    $\vec{u} \cdot (\vec{u} \times \vec{v}) =\begin{bmatrix}
2 \\
-1 \\
4
\end{bmatrix} \cdot
\begin{bmatrix}
-13 \\
2\\
7
\end{bmatrix}
=(2)(−13)+(−1)(2)+(4)(7)=−26−2+28=0$

    Tegak lurus terhadap $\vec{u}$.

2. Cek apakah $\vec{v} \cdot (\vec{u} \times \vec{v}) = 0$:

    $\vec{v} \cdot (\vec{u} \times \vec{v}) =\begin{bmatrix}
3 \\
2 \\
5
\end{bmatrix} \cdot
\begin{bmatrix}
-13 \\
2\\
7
\end{bmatrix}
=(3)(−13)+(2)(2)+(5)(7)=−39+4+35=0$
 
     Tegak lurus terhadap $\vec{v}$.
 
**Kesimpulan:**
Hasil kali silang dari

$\vec{u} =
\begin{bmatrix}
2 \\
-1 \\
4
\end{bmatrix}$ dan $\vec{v} =
\begin{bmatrix}
3 \\
2 \\
5
\end{bmatrix}$

adalah:

$$\vec{u} \times \vec{v} = \begin{bmatrix}
-13 \\
2\\
7
\end{bmatrix}$$

dan hasil ini tegak lurus terhadap kedua vektor.

## Menggunakan determinan untuk menemukan perkalian silang
Pertama, kita membentuk array 3 x 3 seperti yang ditunjukkan di bawah ini.
$$\begin{bmatrix}
\vec{i} & \vec{j} & \vec{k} \\
u_1 & u_2 & u_3 \\
v_1 & v_2 & v_3
\end{bmatrix}$$

$\vec{u} \times \vec{v} =\begin{bmatrix}
u_2 & u_3\\
u_1 & u_3
\end{bmatrix}
\vec{i} -
\begin{bmatrix}
u_1 & u_3\\
u_1 & u_3
\end{bmatrix}
\vec{j} +
\begin{bmatrix}
u_1 & u_2\\
u_1 & u_2
\end{bmatrix}
\vec{k}=$

$= (u_2 v_3 - u_3 v_2) \vec{i} - (u_1 v_3 - u_3 v_1) \vec{j} + (u_1 v_2 - u_2 v_1) \vec{k}$

**Contoh**

$$\left| 
\begin{array}{ccc}
\vec{i} & \vec{j} & \vec{k} \\
2 & -1 & 4 \\
3 & 2 & 5 \\
\end{array}
\right| \rightarrow
\quad \left| 
\begin{array}{cc}
-1 & 4 \\
2 & 5 \\
\end{array}
\right| \vec{i} = (-1)(5) - (4)(2) \, \vec{i} = -13 \, \vec{i}.$$

$$\left| 
\begin{array}{ccc}
\vec{i} & \vec{j} & \vec{k} \\
2 & -1 & 4 \\
3 & 2 & 5 \\
\end{array}
\right| \rightarrow
\quad \left| 
\begin{array}{cc}
2 & 4 \\
3 & 5 \\
\end{array}
\right| \vec{i} = (-2)(5) - (4)(3) \, \vec{j} = 2 \, \vec{j}.$$

$$\left| 
\begin{array}{ccc}
\vec{i} & \vec{j} & \vec{k} \\
2 & -1 & 4 \\
3 & 2 & 5 \\
\end{array}
\right| \rightarrow
\quad \left| 
\begin{array}{cc}
2 & -1 \\
3 & 2 \\
\end{array}
\right| \vec{k} = (2)(2) - (-1)(3) \, \vec{k} = 7 \, \vec{k}.$$

**Sehingga:**

$$\vec{u} \times \vec{v} = \left| 
\begin{array}{ccc}
\vec{i} & \vec{j} & \vec{k} \\
2 & -1 & 4 \\
3 & 2 & 5 \\
\end{array}
\right| 
= 13 \vec{i} - (-2) \vec{j} + 7 \vec{k} 
= \begin{bmatrix}
-13 \\
2 \\
7 \\
\end{bmatrix}$$
## Sifat-sifat Perkalian Silang
1. $\vec{u} \times \vec{v} = -(\vec{v} \times \vec{u})$ *Anticommutative Property*
2. a. $(\vec{u} + \vec{v}) \times \vec{w} = \vec{u} \times \vec{w} + \vec{v} \times \vec{w}$ *Distributive Properties*
b. $\vec{u} \times (\vec{v} + \vec{w}) = \vec{u} \times \vec{v} + \vec{u} \times \vec{w}$
3. ${c}(\vec{u} \times \vec{v}) = (c\vec{u}) \times \vec{v} = \vec{u} \times (c\vec{v})$
4. a. $(\vec{u} \times \vec{v}) \cdot \vec{u} = 0$ *Orthogonality Properties*
b. $(\vec{u} \times \vec{v}) \cdot \vec{v} = 0$
5. $\vec{u} \times \vec{u} = \vec{0}$
6. $\vec{u} \times \vec{0} = \vec{0}$
7. $\vec{u} \cdot (\vec{v} \times \vec{w}) = (\vec{u} \times \vec{v}) \cdot \vec{w}$ *Triple Scalar Product*

### Hasil Kali Silang dan Sudut
Misalkan $\vec{u}$ dan $\vec{v}$ adalah vektor tak nol di $\mathbb{R}^3$. Maka

$\| \vec{u} \times \vec{v} \| = \| \vec{u} \| \| \vec{v} \| \sin(\theta),$

di mana $\theta, 0 \leq \theta \leq \pi$, adalah sudut antara $\vec{u}$ dan $\vec{v}$ 
### Latihan
Misalkan $\vec{u} = \begin{bmatrix} 1 \\ 3 \\ 6 \end{bmatrix}, \quad \vec{v} = \begin{bmatrix} -1 \\ 2 \\ 1 \end{bmatrix}$ Hitung $\|\vec{u} \times \vec{v}\|$

**Langkah 1: Hitung** $\vec{u} \times \vec{v}$
Hasil kali silang dua vektor $\vec{u} = 
\begin{bmatrix} 
u_1 \\ 
u_2 \\ 
u_3 
\end{bmatrix}$ dan $\vec{v} = 
\begin{bmatrix} 
v_1 \\ 
v_2 \\ 
v_3 
\end{bmatrix}$ didefinisikan sebagai:

$\vec{u} \times \vec{v} = \begin{bmatrix}
u_2 v_3 - u_3 v_2 \\
u_3 v_1 - u_1 v_3 \\
u_1 v_2 - u_2 v_1
\end{bmatrix}$

Substitusi nilai $\vec{u}$ dan $\vec{v}$:

$$\vec{u} \times \vec{v} = \begin{bmatrix}
(3)(1) - (6)(2) \\
(6)(-1) - (1)(1) \\
(1)(2) - (3)(-1)
\end{bmatrix} = \begin{bmatrix}
3 - 12 \\
-6 - 1 \\
2 + 3
\end{bmatrix} = \begin{bmatrix}
-9 \\
-7 \\
5
\end{bmatrix}$$

**Langkah 2: Hitung Norma** $\|\vec{u} \times \vec{v}\|$

Norma dari vektor $\vec{w} = \begin{bmatrix} w_1 \\ w_2 \\ w_3 \end{bmatrix}$ adalah:

$$\|\vec{w}\| = \sqrt{w_1^2 + w_2^2 + w_3^2}$$

Sehingga:

$$\|\vec{u} \times \vec{v}\| = \sqrt{(-9)^2 + (-7)^2 + 5^2} = \sqrt{81 + 49 + 25} = \sqrt{155}$$

Jawaban Akhir:

$$\boxed{\sqrt{155}}$$

## Implementasi
### Luas Jajar Genjang
Secara geometri luas jajaran genjang adalah $A = bh$, di mana $b$ adalah panjang alas dan $h$ adalah tinggi jajaran genjang, seperti yang ditunjukkan dalam **Gambar (a)**. Ketika kita mendefinisikan **aturan jajaran genjang penjumlahan vektor**, dua vektor $\vec{u}$ dan $\vec{v}$ membentuk jajaran genjang jika ditarik dari titik awal yang sama, seperti yang ditunjukkan dalam **Gambar (b)**. Trigonometri mengatakan bahwa $h = \| \vec{u} \| \sin(\theta)$, sehingga luas jajaran genjang adalah

$A = \| \vec{u} \| \| \vec{v} \| \sin(\theta) = \| \vec{u} \times \vec{v} \|$

![image](https://hackmd.io/_uploads/BJleBiHQgg.png)

### Contoh
Tentukan luas jajaran genjang yang ditentukan oleh vektor 
$\vec{u} = 
\begin{bmatrix}
2 \\
1
\end{bmatrix}$ dan $\vec{v} = 
\begin{bmatrix}
1 \\
3
\end{bmatrix}$.

![image](https://hackmd.io/_uploads/BkhKIjHmle.png)

Penyelesaiannya adalah dengan memandang $\vec{u}$ dan $\vec{v}$ sebagai vektor di bidang $x-y$ dari $\mathbb{R}^3$, dan menulisnya ulang sebagai:

$\vec{u} = 
\begin{bmatrix}
2 \\
1 \\
0
\end{bmatrix}, \quad \vec{v} = 
\begin{bmatrix}
1 \\
3 \\
0
\end{bmatrix}.$

Kini kita dapat menghitung hasil kali silang. Dengan mudah dapat ditunjukkan bahwa:

$$\vec{u} \times \vec{v} = 
\begin{bmatrix}
0 \\
0 \\
5
\end{bmatrix}.$$

Oleh karena itu, luas jajaran genjang adalah:

$$A = \| \vec{u} \times \vec{v} \| = 5.$$

**Volume Paralelepiped** 
bangun ruang tiga dimensi yang dibentuk oleh enam buah jajar genjang

Dengan menghitung hasil kali silang $\vec{v}$ dan $\vec{w}$, kita mendapatkan vektor yang besarnya sama dengan luas alas. Menghitung produk titik vektor ini dengan $\vec{u}$ menghitung volume parallelepiped!

Oleh karena itu, **volume parallelepiped yang ditentukan oleh vektor** $\vec{u}, \vec{v}$, **dan** $\vec{w}$ **adalah:**

$V = |\vec{u} \cdot (\vec{v} \times \vec{w})|.$
### Contoh
**Menghitung Volume Paralelepiped**

Tentukan volume paralelepiped yang ditentukan oleh vektor $\vec{u} = 
\begin{bmatrix}
1 \\
1 \\
0
\end{bmatrix}$, $\vec{v} = 
\begin{bmatrix}
-1 \\
1 \\
0
\end{bmatrix}$, dan $\vec{w} = 
\begin{bmatrix}
0 \\
1 \\
1
\end{bmatrix}$

Pertama, kita hitung $\vec{v} \times \vec{w} = 
\begin{bmatrix} 
1 \\
1 \\ 
-1 
\end{bmatrix}.$ kemudian $|\vec{u} \cdot (\vec{v} \times \vec{w})| = 
\begin{bmatrix} 
1 \\ 
1 \\ 
0 
\end{bmatrix} \cdot 
\begin{bmatrix} 
1 \\ 
1 \\ 
-1 
\end{bmatrix} 
= 2.$

Oleh karena itu, volume paralelepiped adalah $2$ satuan kubik.

# Tugas
## Buat materi ringkasan perkalian silang  dan kerjakan soal berikut
1. Tentukan luas jajaran genjang yang ditentukan oleh vektor $\vec{u} = \begin{bmatrix} 1 \\ 2 \end{bmatrix},$ dan $\vec{v} = \begin{bmatrix} 2 \\ 1 \end{bmatrix}.$

    **jawaban:**
    
    Gunakan determinan untuk menghitung luas:

$$
\text{Luas} = |\vec{u} \times \vec{v}| = \left| \begin{vmatrix} 1 & 2 \\ 2 & 1 \end{vmatrix} \right| = |(1)(1) - (2)(2)| = |-3| = \boxed{3}
$$

2. Tentukan luas jajaran genjang yang ditentukan oleh vektor $\vec{u} = \begin{bmatrix} 2 \\ 0 \end{bmatrix},$ dan $\vec{v} = \begin{bmatrix} 0 \\ 3 \end{bmatrix}.$

    **Jawaban:**
$$
\text{Luas} = |\vec{u} \times \vec{v}| = \left| \begin{vmatrix} 2 & 0 \\ 0 & 3 \end{vmatrix} \right| = |(2)(3) - (0)(0)| = \boxed{6}
$$
3. Tentukan luas segitiga dengan titik-titik sudut $(0, 0, 0), (1, 3, -1),$ dan  $(2, 1, 1).$

    **Jawaban:**
    * Vektor AB dan AC:

        $\vec{AB} = B - A = (1,3,-1), \quad \vec{AC} = C - A = (2,1,1)$

    * Hitung cross product $\vec{AB} \times \vec{AC}$:
        
        $\vec{AB} \times \vec{AC} =
\begin{vmatrix}
\hat{i} & \hat{j} & \hat{k} \\
1 & 3 & -1 \\
2 & 1 & 1 \\
\end{vmatrix} = 
\hat{i}(3\cdot1 - (-1)\cdot1) - \hat{j}(1\cdot1 - (-1)\cdot2) + \hat{k}(1\cdot1 - 3\cdot2) = \hat{i}(4) - \hat{j}(3) + \hat{k}(-5)$

    * Panjang vektor hasil cross product:

        $|\vec{AB} \times \vec{AC}| = \sqrt{4^2 + (-3)^2 + (-5)^2} = \sqrt{16 + 9 + 25} = \sqrt{50}$

    * Luas segitiga: 

        $\text{Luas} = \frac{1}{2} |\vec{AB} \times \vec{AC}| = \frac{1}{2} \sqrt{50} = \boxed{\frac{\sqrt{50}}{2}} \approx \boxed{3.54}$

4.  Tentukan luas segitiga dengan titik-titik sudut $(5, 2, -1), (3, 6, 2),$ dan  $(1, 0, 4).$

    **Jawaban:**
    * Vektor AB dan AC:

        $\vec{AB} = B - A = (-2, 4, 3), \quad \vec{AC} = C - A = (-4, -2, 5)$

    * Cross product:
    
        $\vec{AB} \times \vec{AC} =
\begin{vmatrix}
\hat{i} & \hat{j} & \hat{k} \\
-2 & 4 & 3 \\
-4 & -2 & 5 \\
\end{vmatrix}
= \hat{i}(4 \cdot 5 - 3 \cdot (-2)) - \hat{j}(-2 \cdot 5 - 3 \cdot (-4)) + \hat{k}(-2 \cdot (-2) - 4 \cdot (-4)$

    * Panjang vektor hasil cross product:
    
        $\sqrt{26^2 + (-2)^2 + 20^2} = \sqrt{676 + 4 + 400} = \sqrt{1080}$
    * Luas segitiga:
    
        $\text{Luas} = \frac{1}{2} \sqrt{1080} = \boxed{\frac{\sqrt{1080}}{2}} \approx \boxed{16.43}$
